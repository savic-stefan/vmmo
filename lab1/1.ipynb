```python
import numpy as np
import matplotlib.pyplot as plt
xx = np.linspace(-2, 4, 101)
plt.plot(xx, norm.pdf(xx, loc=1), label='loc=%s, scale=%s'%(1, 1))
plt.plot(xx, norm.pdf(xx, loc=1, scale=0.5),
label='loc=%s, scale=%s'%(1, 0.5))
plt.grid()
plt.legend(loc='best')
```
