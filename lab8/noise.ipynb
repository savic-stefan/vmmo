```python
import numpy as np
import matplotlib.pyplot as plt
noisy_image= np.loadtxt("noisy_image.csv",  delimiter=",")

plt.imshow(noisy_image, cmap='gray')
plt.axis('off')
plt.show()
#код генерации шума
def add_impulse_noise(image, p=0.2):
    noisy_image = image.copy()
    num_pixels = image.size
    num_noisy = int(p * num_pixels)
    coords = np.random.choice(num_pixels, num_noisy, replace=False)
    noisy_image.flat[coords] = np.random.choice([0, 1], size=num_noisy)  # Чёрный или белый шум
    return noisy_image
```
